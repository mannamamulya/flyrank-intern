## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The target is whether a content page is declining in search performance. In the starter dataset, I use the trend_direction column as the target. Pages with trend_direction = "down" are labeled as declining, while all other pages are labeled as not declining. In the starter dataset this label is defined from the observed trend percentage, so it is suitable for learning the workflow, but future work should use a label measured in a later time window.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

df["is_declining"] = (df["trend_direction"] == "down").astype(int)

df[["trend_direction", "is_declining"]].head()
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

The success metric is Precision@50. This measures how many of the top 50 pages recommended by the model are actually declining. A higher Precision@50 means the recommendations are more useful because content teams spend their time reviewing the right pages.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
print("Declining pages:", df["is_declining"].sum())
print("Total pages:", len(df))
print("Declining rate:", round(df["is_declining"].mean(), 3))
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

However, search performance depends on many factors working together, including impressions, rankings, CTR, engagement, and content characteristics. A machine learning model can learn these patterns from historical data and provide better decision support than a simple hand-written rule.
